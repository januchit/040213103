## ข้อ 1 : สี่เหลี่ยมผืนผ้า

รับ **ความกว้าง** และ **ความยาว** ของสี่เหลี่ยมผืนผ้าจากผู้ใช้ด้วย `input()` แล้วแสดง **พื้นที่** และ **เส้นรอบรูป**

$$\text{พื้นที่} = \text{กว้าง} \times \text{ยาว} \qquad \text{เส้นรอบรูป} = 2 \times (\text{กว้าง} + \text{ยาว})$$

**ตัวอย่างการทำงาน**
```
ความกว้าง: 4
ความยาว: 7
พื้นที่ = 28.0
เส้นรอบรูป = 22.0
```

````python
# ✏️ เขียนโค้ดของคุณที่นี่
# คำแนะนำ: input() คืนค่าเป็นข้อความ (str) ต้องแปลงเป็นตัวเลขด้วย float() ก่อนคำนวณ
````

## ข้อ 2 : ค่าเฉลี่ย

รับ **คะแนน 3 วิชา** จากผู้ใช้ (เป็นทศนิยมได้) แล้วแสดง **ค่าเฉลี่ย** เป็น **ทศนิยม 2 ตำแหน่ง**

**ตัวอย่างการทำงาน**
```
คะแนนวิชาที่ 1: 78.5
คะแนนวิชาที่ 2: 82
คะแนนวิชาที่ 3: 90.25
ค่าเฉลี่ย = 83.58
```

````python
# ✏️ เขียนโค้ดของคุณที่นี่
# คำแนะนำ: ใช้ f-string เช่น f"{ค่า:.2f}" เพื่อแสดงทศนิยม 2 ตำแหน่ง
````
